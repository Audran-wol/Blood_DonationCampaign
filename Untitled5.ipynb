```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import datetime as dt
```

```python
df=pd.read_csv('/content/Final_Cleaned_and_Merged_DataFrame.csv')
```

```python
df['date_de_naissance'] = pd.to_datetime(df['date_de_naissance'], errors='coerce')

# Calculate age
# Use dt.datetime.now() to get the current datetime object
current_year = dt.datetime.now().year
df['age'] = current_year - df['date_de_naissance'].dt.year

# Handle any invalid dates (like 2089, 2092 which are clearly errors)
# We'll cap ages at 100 and set minimum at 15 (assuming blood donors must be at least 15)
df['age'] = df['age'].clip(15, 100)

# Show the result
print(df[['date_de_naissance', 'age']].head())
```

```python
df['age'] = df['age'].fillna(-1).astype(int)
```

```python
from geopy.geocoders import Nominatim
import time

# Initialize geocoder
geolocator = Nominatim(user_agent="blood_donation_analysis")

# Function to get coordinates
def get_coordinates(location):
    try:
        location_info = geolocator.geocode(location + ", Douala, Cameroon")
        if location_info:
            return location_info.latitude, location_info.longitude
        else:
            return None, None
    except:
        return None, None

# Since geocoding takes time, we'll create a mapping of known areas
# This is a sample - you should expand this with actual coordinates for Douala areas
known_locations = {
    'Douala 1': (4.0483, 9.7043),
    'Douala 2': (4.0511, 9.7181),
    'Douala 3': (4.0658, 9.7674),
    'Douala 4': (4.0833, 9.7333),
    'Douala 5': (4.0667, 9.7000),
    'Yaounde': (3.8480, 11.5021),
    'Buea': (4.1667, 9.2333),
    'Edea': (3.8000, 10.1333),
    'Non précisé': (4.0511, 9.7678)  # Default to Douala center
}

# Apply the mapping
df['latitude'] = df['residence_area'].map(lambda x: known_locations.get(x, known_locations['Non précisé'])[0])
df['longitude'] = df['residence_area'].map(lambda x: known_locations.get(x, known_locations['Non précisé'])[1])

# Show the result
print(df[['residence_area', 'latitude', 'longitude']].head())
```
